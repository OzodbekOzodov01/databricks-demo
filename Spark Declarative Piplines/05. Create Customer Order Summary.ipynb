```sql
create or refresh materialized view circuitbox.lakehouse.gold_customer_order_summary
as
select c.customer_id,
       c.customer_name,
       c.date_of_birth,
       c.telephone,
       c.email,
       a.address_line_1,
       a.city,
       a.state,
       a.postcode,
       count(distinct o.order_id) as total_orders,
       sum(o.item_quantity) as total_items_ordered,
       sum(o.item_price * o.item_quantity) as total_order_amount
from circuitbox.lakehouse.silver_customers c
join circuitbox.lakehouse.silver_addresses a on c.customer_id = a.customer_id
join circuitbox.lakehouse.silver_orders o on c.customer_id = o.customer_id
where a.__END_AT is null
group by all;
```
