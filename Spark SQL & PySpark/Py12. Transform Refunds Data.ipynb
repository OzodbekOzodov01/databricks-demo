```python
df = spark.table('gizmobox.bronze.py_refunds')
display(df)
```

```python
from pyspark.sql import functions as F
df_extract_refunds = (
                        df
                        .select(
                                'refund_id',
                                'payment_id',
                                 F.date_format("refund_timestamp", "yyyy-MM-dd").cast('date').alias('refund_date'),
                                 F.date_format("refund_timestamp", "HH:mm:ss").alias('refund_time'),
                                'refund_amount',
                                F.split('refund_reason', ':')[0].alias('refund_reason'),
                                F.split('refund_reason', ':')[1].alias('refund_sourse')
                        )
)
display(df_extract_refunds)
```

```python
df_extract_refunds.writeTo('gizmobox.silver.py_refunds').createOrReplace()
```
