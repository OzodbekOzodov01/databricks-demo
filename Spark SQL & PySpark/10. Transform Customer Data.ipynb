```sql
create or replace temporary view v_customers_distinct as
select distinct * from gizmobox.bronze.v_customers
where customer_id is not null
order by customer_id
```

```sql
with cte as (
    select customer_id,
        max(created_timestamp) as max_created_timestamp
    from v_customers_distinct
    group by customer_id
)
select 
    cast(d.created_timestamp as timestamp) as created_timestamp,
    d.customer_id,
    d.customer_name,
    cast(d.date_of_birth as date) as date_of_birth,
    d.email,
    cast(d.member_since as date) as member_since,
    d.telephone
from v_customers_distinct d
inner join cte c
on d.customer_id = c.customer_id
and d.created_timestamp = c.max_created_timestamp
```

```sql
create table gizmobox.silver.customers
as 
with cte as (
    select customer_id,
        max(created_timestamp) as max_created_timestamp
    from v_customers_distinct
    group by customer_id
)
select 
    cast(d.created_timestamp as timestamp) as created_timestamp,
    d.customer_id,
    d.customer_name,
    cast(d.date_of_birth as date) as date_of_birth,
    d.email,
    cast(d.member_since as date) as member_since,
    d.telephone
from v_customers_distinct d
inner join cte c
on d.customer_id = c.customer_id
and d.created_timestamp = c.max_created_timestamp
```
