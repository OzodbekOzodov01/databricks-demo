```python
df_customers = spark.table('gizmobox.silver.py_customers')
df_addresses = spark.table('gizmobox.silver.py_addresses')
```

```python
df_customer_address = (
    df_customers
    .join(df_addresses, 'customer_id', 'inner')
    .select(
        'customer_id',
        'customer_name',
        'email',
        'date_of_birth',
        'member_since',
        'telephone',
        'shipping_address_line_1',
        'shipping_city',
        'shipping_state',
        'shipping_postcode',
        'billing_address_line_1',
        'billing_city',
        'billing_state',
        'billing_postcode'
    )
)
display(df_customer_address)
```

```python
df_customer_address.writeTo('gizmobox.gold.py_customer_address').createOrReplace()
```
