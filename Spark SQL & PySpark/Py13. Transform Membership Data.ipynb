```python

df = spark.table('gizmobox.bronze.py_memberships')
display(df)
```

```python
from pyspark.sql import functions as F
df_extract_membership = (
                        df
                        .select(
                                F.regexp_extract('path', '.*/([0-9]+)\\.png$', 1).alias('customer_id'),
                                df.content.alias('membership_card')
                        )
)
display(df_extract_membership)
```

```python
df_extract_membership.writeTo('gizmobox.silver.py_memberships').createOrReplace()
```
