```sql
create table gizmobox.silver.payments
as
select 
  payment_id,
  order_id,
  date_format(payment_timestamp, 'yyyy-MM-dd')::date as payment_date,
  date_format(payment_timestamp, 'HH:mm:ss') as payment_time,
  case payment_status
    when 1 then 'Success'
    when 2 then 'Pending'
    when 3 then 'Canceled'
    when 4 then 'Failed'
  end as payment_status,
  payment_method  
from gizmobox.bronze.payments
```
