```sql
create table if not exists gizmobox.silver.refunds
as
select 
    refund_id,
    payment_id,
    date_format(refund_timestamp, 'yyyy-MM-dd')::date as refund_date,
    date_format(refund_timestamp, 'HH:mm:ss') as refund_time,
    refund_amount,
    split(refund_reason, ':')[0] as refund_reason,
    split(refund_reason, ':')[1] as refund_sourse
from gizmobox.bronze.refunds
```
