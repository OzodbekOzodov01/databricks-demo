```python
df = spark.table('gizmobox.silver.py_order')
display(df)
```

```python
from pyspark.sql import functions as F
df_monthly_order = (
    df
    .groupBy(F.date_format('transaction_timestamp', 'yyyy-MM').alias('order_month'), "customer_id")
    .agg(
        F.count_distinct('order_id').alias('total_orders'),
        F.sum('quantity').alias('total_item_bought'),
        F.sum(df.price * df.quantity).alias('total_amount')    
    )
)
display(df_monthly_order)
```

```python
df_monthly_order.writeTo('gizmobox.gold.py_order_monthly_summary').createOrReplace()
```
