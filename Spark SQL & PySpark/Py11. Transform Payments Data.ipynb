```python
df = spark.table('gizmobox.bronze.py_payments')
display(df)
```

```python
from pyspark.sql import functions as F
df_extracted_payment = (
                        df
                        .select("payment_id",
                                "order_id",
                                F.date_format("payment_timestamp", "yyyy-MM-dd").cast('date').alias('payment_date'),
                                F.date_format("payment_timestamp", "HH:mm:ss").alias('payment_time'),
                                F.when(df.payment_status == 1, "Success")
                                .when(df.payment_status == 2, 'Pending')
                                .when(df.payment_status == 3, 'Canceled')
                                .when(df.payment_status == 4, 'Failed')
                                .alias('payment_status'),
                                "payment_method"
                                )
)
display(df_extracted_payment)
```

```python
df_extracted_payment.writeTo("gizmobox.silver.py_payments").createOrReplace()
```
