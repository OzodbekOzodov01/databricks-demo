```python
df = spark.table('gizmobox.bronze.py_addresses')
display(df)
```

```python
from pyspark.sql.functions import max
df_extracted_addresses = (
                        df
                        .groupBy('customer_id')
                        .pivot("address_type", ["shipping", "billing"])
                        .agg(
                            max("address_line_1").alias("address_line_1"),
                            max("city").alias("city"),
                            max("state").alias("state"),
                            max("postcode").alias("postcode")
                        )
)
display(df_extracted_addresses)
```

```python
df_extracted_addresses.writeTo('gizmobox.silver.py_addresses').createOrReplace()
```
